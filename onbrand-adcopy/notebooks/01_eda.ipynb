# OnBrand AdCopy - Exploratory Data Analysis
Reproducing the three capstone figures from cleaned data.

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_prep import load_and_clean
from src.features import add_features

df = load_and_clean()
df = add_features(df)
print(f'Dataset: {len(df)} rows')
df.head()
```

## Figure 1: Word Count vs CTR Proxy

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df['word_count'], df['ctr_proxy'], alpha=0.3, s=10)
z = np.polyfit(df['word_count'], df['ctr_proxy'], 1)
p = np.poly1d(z)
x_line = np.linspace(df['word_count'].min(), df['word_count'].max(), 100)
ax.plot(x_line, p(x_line), 'r--', linewidth=2, label=f'OLS trend (slope={z[0]:.3f})')
ax.set_xlabel('Word Count')
ax.set_ylabel('CTR Proxy (%)')
ax.set_title('Word Count vs. CTR Proxy')
ax.legend()
plt.tight_layout()
plt.savefig('../reports/eda_figures/word_count_vs_ctr.png', dpi=150)
plt.show()
```

## Figure 2: Mean CTR by Question and Urgency

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

q_means = df.groupby('has_question')['ctr_proxy'].mean()
axes[0].bar(['No Question', 'Has Question'], [q_means.get(0, 0), q_means.get(1, 0)], color=['#4C72B0', '#DD8452'])
axes[0].set_ylabel('Mean CTR Proxy (%)')
axes[0].set_title('Mean CTR by Has Question')

u_means = df.groupby('has_urgency')['ctr_proxy'].mean()
axes[1].bar(['No Urgency', 'Has Urgency'], [u_means.get(0, 0), u_means.get(1, 0)], color=['#55A868', '#C44E52'])
axes[1].set_ylabel('Mean CTR Proxy (%)')
axes[1].set_title('Mean CTR by Has Urgency')

plt.tight_layout()
plt.savefig('../reports/eda_figures/ctr_by_features.png', dpi=150)
plt.show()
```

## Figure 3: Correlation Heatmap

```python
model_features = ['char_count', 'word_count', 'has_question', 'has_exclamation',
                  'urgency_count', 'sentiment_score', 'toxicity_score', 'ctr_proxy']
corr = df[model_features].corr()
fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r', center=0, ax=ax, square=True)
ax.set_title('Feature Correlation Heatmap')
plt.tight_layout()
plt.savefig('../reports/eda_figures/correlation_heatmap.png', dpi=150)
plt.show()
```
